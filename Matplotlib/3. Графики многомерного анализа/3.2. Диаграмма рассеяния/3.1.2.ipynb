```python
from sklearn.datasets import load_breast_cancer

# Загрузка датасета
cancer = load_breast_cancer()
data = cancer.data

mean_radius = data[:, 0]
mean_area = data[:, 3]
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.scatter(mean_radius, mean_area, alpha=0.6)
plt.xlabel('Средний радиус')
plt.ylabel('Средняя площадь')
plt.title('Зависимость площади опухоли от радиуса')
plt.grid()

coef = np.polyfit(mean_radius, mean_area, 1)
trend = np.poly1d(coef)
square = np.random.randint(0, 30, 60)
x_smooth = np.linspace(square.min(), square.max(), 100)
plt.plot(x_smooth, trend(x_smooth),
         color='red', linewidth=2, linestyle='--')

plt.show()
```

```python
target = cancer.target

radius_malignant = mean_radius[target == 0]
area_malignant = mean_area[target == 0]
radius_benign = mean_radius[target == 1]
area_benign = mean_area[target == 1]
```

```python
plt.scatter(radius_malignant, area_malignant, marker='o', c='red', label='Злокачественные')
plt.scatter(radius_benign, area_benign, marker='s', c='blue', label='Доброкачественные')
plt.xlabel('Средний радиус')
plt.ylabel('Средняя площадь')
plt.title('Сравнение размеров опухолей по типам')
plt.show()
```

```python
mean_radius = data[:, 0]       # средний радиус
mean_area = data[:, 3]         # средняя площадь
mean_smoothness = data[:, 4]   # средняя гладкость
mean_concavity = data[:, 6]    # средняя вогнутость
```

```python
scatter = plt.scatter(mean_radius, mean_area, c=mean_smoothness, s=mean_concavity*200, alpha=0.6)
plt.xlabel('Средний радиус')
plt.ylabel('Средняя площадь')
plt.colorbar(scatter)
plt.title("Зависимость четырех признаков опухолей")
plt.show()
```
