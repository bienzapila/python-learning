```python
import numpy as np
from sklearn.datasets import load_digits

digits = load_digits()
target = digits.target

unique, counts = np.unique(target, return_counts=True)
print("Цифры:", unique)
print("Количество:", counts)
```

```python
import matplotlib.pyplot as plt

plt.bar(unique, counts)
plt.title('Распределение цифр в датасете')
plt.xlabel('Цифра')
plt.ylabel('Количество')
plt.grid(axis='y')
plt.show()
```

```python
data = digits.data
mean_brightness = [data[target == i].mean() for i in range(10)]

print(mean_brightness)
```

```python
plt.bar(unique, mean_brightness, color='coral')
plt.xlabel('Цифра')
plt.ylabel('Средняя яркость (0-16)')
plt.title('Средняя яркость изображения по цифрам')
plt.grid(axis='y', alpha=0.4)
plt.xticks(unique)
plt.show()
```
