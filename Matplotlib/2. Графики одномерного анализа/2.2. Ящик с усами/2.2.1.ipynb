```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Создаем данные с нужными характеристиками
data = np.concatenate([
    np.random.normal(50, 4, 80),      # основная масса данных (80 точек)
    np.random.normal(35, 4, 3),       # группа низких значений (5 точек)
    np.random.normal(65, 4, 3)        # группа высоких значений (5 точек)
])

# Добавляем несколько явных выбросов

box = plt.boxplot(data,
                  vert=True,  # вертикальная ориентация
                  patch_artist=True,  # закрашиваем ящики
                  notch=True,  # включаем выемку у медианы

                  boxprops=dict(facecolor='lightblue',
                                edgecolor='navy',
                                linewidth=2,
                                linestyle='-'),

                  # Настройка медианы
                  medianprops=dict(color='red',
                                   linewidth=3,
                                   linestyle='-'),

                  # Настройка усов
                  whiskerprops=dict(color='darkgreen',
                                    linewidth=2,
                                    linestyle='--'),

                  # Настройка засечек на усах
                  capprops=dict(color='purple',
                                linewidth=2,
                                linestyle='-'),

                  # Настройка выбросов
                  flierprops=dict(marker='o',
                                  markerfacecolor='orange',
                                  markersize=8,
                                  alpha=0.7,
                                  markeredgecolor='darkred',
                                  markeredgewidth=1.5))

plt.show()
```
