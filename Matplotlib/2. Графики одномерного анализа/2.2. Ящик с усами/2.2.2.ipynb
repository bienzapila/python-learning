```python
from sklearn.datasets import load_diabetes

# Загрузка данных
diabetes = load_diabetes()
data = diabetes.data
bmi = data[:, 2]
```

```python
import matplotlib.pyplot as plt

plt.boxplot(bmi, patch_artist=True, boxprops=dict(facecolor='lightblue'), medianprops=dict(linewidth=2, color='red'), whiskerprops=dict(color='blue', linestyle='--' ))
plt.ylabel('Индекс массы тела')
plt.title('Распределение индекса массы тела у пациентов с диабетом')
plt.show()
```

```python
plt.boxplot(
    bmi,
    vert=False,
    showmeans=True,
    flierprops=dict(markerfacecolor='orange', markersize=8)
)
plt.xlabel('Индекс массы тела')
plt.title('Распределение массы тела у пациентов с диабетом')
plt.show()
```

```python
import numpy as np

age = diabetes.data[:, 0] 
median_age = np.median(age)
young_patients = target[age <= median_age]
old_patients = target[age > median_age]
```

```python
box = plt.boxplot([young_patients, old_patients], tick_labels=['Молодые пациенты', 'Старые пациенты'], patch_artist=True, notch=True)
box['boxes'][0].set_facecolor('lightgreen')
box['boxes'][1].set_facecolor('lightblue')
plt.title('Сравнение прогресии диабета в разных возрастных группах')
plt.ylabel('Прогрессия диабета')
plt.show()
```
