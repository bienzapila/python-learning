Пакет numpy - установка и первое знакомство

```python
import numpy as np
```

```python
array = np.array([1, 2, 3, 4, 5], dtype=list)
array.dtype
```

```python
a = np.array([1, True])
a
```

```python
array[0] = 10
array[0]
```

```python
array[0] = 'a'
array.dtype
```

```python
a = np.array([i for i in range(1, 10)])
a[2]
```

```python
a[[1,1,1,1,1]]
```

```python
a[[True for i in range(1, 10)]]
```

```python
a = np.array([1, 2, 7, 8, 9, 3, 6, 7, 3])
b = a.reshape(3, 3)
b
```

```python
b[1][2]
```
