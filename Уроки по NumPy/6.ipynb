Объединение и разделение массивов

```python
import numpy as np
```

```python
a = np.array([(1, 2), (3, 4)])
b = np.array([(5, 6), (7, 8)])
```

```python
np.hstack((a, b))
```

```python
np.vstack((a, b))
```

```python
a = np.fromstring('1 2 3 4', sep=' ')
b = np.fromstring('5 6 7 7', sep = ' ')
```

```python
np.column_stack((a, b))
```

```python
np.row_stack((a, b))
```

```python
a = np.arange(1, 13)
b = np.arange(13, 26)
```

```python
a.resize(3, 3, 2, refcheck=False)
b.resize(3, 3, 2, refcheck=False)
```

```python
np.concatenate((a, b), axis=0)
```

```python
np.r_[[1, 2, 3], 4, 5]
```

```python
np.r_[1:9, 90, 100]
```

```python
np.c_[1:5]
```

```python
a = np.arange(10)
```

```python
np.hsplit(a, 2)
```

```python
a = np.ones((3, 3))
a
```

```python
np.vsplit(a, 3)
```
